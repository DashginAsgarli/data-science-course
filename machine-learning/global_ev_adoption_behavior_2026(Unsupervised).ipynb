```python
df_cluster = df.drop(columns=[
    'ev_adoption_likelihood',
    'monthly_energy_consumption_kwh',
    'monthly_charging_cost'
])
```

```python
df_cluster.dtypes
```

```python
df_cluster = pd.get_dummies(
    df_cluster,
    columns=['education_level', 'city_type', 'current_vehicle_type'],
    drop_first=True
)
```

```python
df_cluster = df_cluster.fillna(df_cluster.median(numeric_only=True))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_cluster = scaler.fit_transform(df_cluster)
```

```python
from sklearn.cluster import KMeans

inertias = []

for i in range(1,10):
  kmeans = KMeans(n_clusters=i)
  kmeans.fit(df_cluster)
  inertias.append(kmeans.inertia_)
print(inertias)
```

```python
plt.plot(range(1,10), inertias, marker = 'o')
plt.title('Elbow Method')
plt.xlabel('Number of cluster')
plt.ylabel('Inertias')
plt.show()
```

```python
from sklearn.decomposition import PCA

kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(df_cluster)

pca = PCA(n_components=2)
principal_components = pca.fit_transform(df_cluster)

plt.scatter(principal_components[:,0], principal_components[:,1],c = kmeans.labels_)
plt.show()
```
