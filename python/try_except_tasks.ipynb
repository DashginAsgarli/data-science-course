```python
try:
    x = int(input("Enter a number: "))
    print(10 / x)

except ZeroDivisionError:
    print("You cannot divide by zero.")

except ValueError:
    print("Please enter a number. You cannot divide by a letter.")

finally:
    print("Process finished.")
```

```python
data = ['10', '20', '12', '34', '0', 'abs']

for item in data:
    try:
        number = int(item)
        result = 100 / number
        print(f"Result for {item}: ", result)

    except ValueError:
        print(f"{item} is not a number")

    except ZeroDivisionError:
        print("Division by zero is not allowed")

    finally:
        print("-------------------")
```

```python
def calc(par, d1, d2):
    try:
        d1 = int(d1)
        d2 = int(d2)

        if par == "+":
            return d1 + d2

        elif par == "*":
            return d1 * d2

    except ValueError:
        return "You cannot perform calculations with strings. Please enter numbers."


print(
    calc(
        input("Enter the operator: "),
        input("Enter the first number: "),
        input("Enter the second number: ")
    )
)
```
