### Sual-1. Rectangle Class

**Sual:**
`Rectangle` adlı class yaradın. Class-da düzbucaqlının `width` (en) və `height` (hündürlük) atributları olsun.

Aşağıdakı metodları yaradın:

* `area()` — düzbucaqlının sahəsini hesablasın.
* `perimeter()` — düzbucaqlının perimetrini hesablasın.
* `is_square()` — düzbucaqlının kvadrat olub-olmadığını yoxlasın.
* `__str__()` — düzbucaqlının eni və hündürlüyünü göstərən məlumat qaytarsın.

**Input:**

```python
a = Rectangle(5, 6)

print(a.area())
print(a.perimeter())
print(a.is_square())
print(a.__str__())
```

**Output:**

```text
30
22
False
Düzbucaqlı: en=5, hündürlük=6
```

````python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height

    def __str__(self):
        return f"Düzbucaqlı: en={self.width}, hündürlük={self.height}"

a = Rectangle(5 ,6)
print(a.area())
print(a.perimeter())
print(a.is_square())
print(a.__str__())
````

### Sual-2. Student Class

**Sual:**
`Student` adlı class yaradın. Class-da tələbənin `first_name`, `last_name` və qiymətlərini (`grades`) saxlayan atributlar olsun.

Aşağıdakı metodları yaradın:

* `add_grade()` — tələbənin qiymətlər siyahısına yeni qiymət əlavə etsin.
* `average()` — tələbənin qiymətlərinin ortalamasını hesablasın. Əgər qiymət yoxdursa, `0` qaytarsın.
* `full_name()` — tələbənin adını və soyadını birləşdirib qaytarsın.
* `__str__()` — tələbənin adını, soyadını və ortalamasını göstərsin.

**Input:**

```python
student1 = Student("Dashqin", "Asgarli")

student1.add_grade(80)
student1.add_grade(75)
student1.add_grade(91)

print(student1.average())
print(student1.full_name())
print(student1.__str__())
```

**Output:**

```text
82.0
Dashqin Asgarli
Tələbə: Dashqin Asgarli, Ortalama: 82.0
```

````python
class Student:

    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def __str__(self):
        return f"Tələbə: {self.full_name()}, Ortalama: {self.average()}"

student1 = Student("Dashqin" , "Asgarli")
student1.add_grade(80)
student1.add_grade(75)
student1.add_grade(91)
print(student1.average())
print(student1.full_name())
print(student1.__str__())
````

### Sual-3. Book və Library Class

**Sual:**
`Book` adlı class yaradın. Class-da kitabın `title` (başlıq), `author` (müəllif) və `year` (nəşr ili) atributları olsun.

`info()` metodu kitab haqqında məlumatı qaytarsın.

Daha sonra `Library` adlı class yaradın. Kitabxanada kitabları saxlamaq üçün `books` siyahısından istifadə edin.

Aşağıdakı metodları yaradın:

* `add_book()` — kitabı kitabxanaya əlavə etsin.
* `list_books()` — kitabxanadakı bütün kitabları göstərsin.
* `find_by_author()` — verilmiş müəllifə aid kitabları tapsın.
* `find_by_year()` — verilmiş ildə nəşr olunmuş kitabları tapsın.
* `remove_book()` — başlığı verilmiş kitabı kitabxanadan silsin. Kitab tapılmadıqda `"Book not found"` yazdırsın.

**Input:**

```python
book1 = Book("1984", "George Orwell", 1949)
book2 = Book("The Hobbit", "J.R.R. Tolkien", 1937)
book3 = Book("Animal Farm", "George Orwell", 1945)

library = Library()

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.list_books()

books_by_orwell = library.find_by_author("George Orwell")
for book in books_by_orwell:
    print(book.info())

books_1945 = library.find_by_year(1945)
for book in books_1945:
    print(book.info())

library.remove_book("1984")
library.list_books()
```

**Output:**

```text
1984 by George Orwell (1949)
The Hobbit by J.R.R. Tolkien (1937)
Animal Farm by George Orwell (1945)

1984 by George Orwell (1949)
Animal Farm by George Orwell (1945)

Animal Farm by George Orwell (1945)

The Hobbit by J.R.R. Tolkien (1937)
Animal Farm by George Orwell (1945)
```

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def info(self):
        return f"{self.title} by {self.author} ({self.year})"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def list_books(self):
        for book in self.books:
            print(book.info())

    def find_by_author(self, author):
        result = []
        for book in self.books:
            if book.author == author:
                result.append(book)
        return result

    def find_by_year(self, year):
        result = []
        for book in self.books:
            if book.year == year:
                result.append(book)
        return result

    def remove_book(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return
        print("Book not found")
        


book1 = Book("1984", "George Orwell", 1949)
book2 = Book("The Hobbit", "J.R.R. Tolkien", 1937)
book3 = Book("Animal Farm", "George Orwell", 1945)

library = Library()

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.list_books()

books_by_orwell = library.find_by_author("George Orwell")
for book in books_by_orwell:
    print(book.info())

books_1945 = library.find_by_year(1945)
for book in books_1945:
    print(book.info())

library.remove_book("1984")
library.list_books()
````

Constructor vasitəsilə avtomobilin aşağıdakı məlumatlarını qəbul edin:

* `start()` — avtomobilin işə düşdüyünü göstərsin.
* `get_wholes_count()` — class-da olan ümumi `wholes` sayını qaytarsın.
* `info()` — `"Cars"` məlumatını göstərsin.
* `create_electric_car()` — marka, model və rəng qəbul edərək elektrik avtomobili yaratsın.

**Input:**

```python
car1 = Cars("BMW", "XS", "Black", "petrol")

print(car1.model)
print(Cars.get_wholes_count())
Cars.info()
```

4

### Sual-5. Circle Class

**Sual:**
`Circle` adlı class yaradın. Class-da `radius` atributu olmalıdır. Radiusun dəyərini dəyişmək üçün bir metod və dairənin sahəsini hesablamaq üçün başqa bir metod yaradın.

**Input:**

```python
c = Circle(5)
print(c.area())

c.set_radius(10)
print(c.area())
```

**Output:**

```text
78.5
314.0
```

````python
import math
class Circle ():
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return round(math.pi ,2) * self.radius **2
    
    def set_radius(self , new_radius):
        self.radius = new_radius

c = Circle(5) 
print(c.area()) 
c.set_radius(10) 
print(c.area())
````

### Sual-6. BankAccount Class

**Sual:**
`BankAccount` adlı class yaradın. Class-da bank hesabının `balance` (balans) atributu olsun.

Aşağıdakı metodları yaradın:

* `deposit()` — hesaba müəyyən məbləğdə pul əlavə etsin.
* `withdraw()` — hesabdan müəyyən məbləğdə pul çıxarsın.
* `get_balance()` — hesabın cari balansını qaytarsın.

**Input:**

```python
acc = BankAccount(1000)

acc.deposit(500)
acc.withdraw(300)

print(acc.get_balance())
```

**Output:**

```text
1200
```

````python
class BankAccount():
    def __init__(self,balance):
        self.balance = balance
        
    def deposit (self , amount) :
        self.balance += amount
    
    def withdraw (self , amount):
        self.balance -= amount
    
    def get_balance(self):
        return self.balance
acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(300)
print(acc.get_balance())
````
