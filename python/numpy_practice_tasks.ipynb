#### 3) Bir massiv yaradın, həmin massivdə maximum və minimum dəyərlərin indekslərini tapan proqram yazın. Nəticələri çap edin.

```python
arr = np.array([5, -3, 2, 18, 9, 3, 41])

max_idx = np.argmax(arr)
min_idx = np.argmin(arr)
```

```python
print(f'Maximum dəyərin indeksi: {max_idx}')
print(f'Minimum dəyərin indeksi: {min_idx}')
```

#### 6) Bir matris yaradın. Həmin matrisin natural, 10 əsaslı və 2 əsaslı logarifmləri hesablayan proqram yazın. Nəticələri çap edin.

```python
matris = np.array([1, 10, 100])

log_e = np.log(matris)
log_10 = np.log10(matris)
log_2 = np.log2(matris)
```

```python
print(f'{log_e}\n {log_10}\n {log_2}')
```

#### 7) Bir matris yaradın və onun determinantını hesablayan proqram yazın. Nəticəni çap edin.

```python
matrix = np.array([
    [3, 2],
    [3, 4]])

det = np.linalg.det(matrix).round(2)
```

```python
det
```

#### 9) İki matris yaradın və onların kovariasiya matrisini hesablayan proram yazın. Nəticəni çap edin.

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]])

B = np.array([
    [2, 4, 6],
    [1, 3, 5]])

cov_mtrx = np.cov(A, B)
```

```python
cov_mtrx
```
