```python
def arrayPairSum(nums):
    nums.sort()
    return sum(nums[::2])
print(arrayPairSum([6,2,6,5,1,2]))
```

```python
def ops(lists):
    my_list =[]
    
    for item in lists:
        if item.lstrip("-").isdigit():
            my_list.append(int(item))
        elif item == "C":
            my_list.pop()
        elif item == "D":
            my_list.append(2 * my_list[-1])
        elif item == "+":
            my_list.append(my_list[-1] + my_list[-2])
    return sum(my_list)

print(ops(["1","C"]))    
```

```python
def numJewelsInStones(jewels, stones):
    total =0
    for item in jewels:
        if item in stones:
            total += stones.count(item)
    return total       
print(numJewelsInStones("aA" , "aAAbbbb"))
```

```python
def removeOuterParentheses(s):
    result = []
    balance = 0
    
    for ch in s:
        if ch == '(':
            if balance > 0:
                result.append(ch)
            balance += 1
        else:
            balance -= 1
            if balance > 0:
                result.append(ch)
    return "".join(result)
```

```python
def min_max(my_list):
    min_eded = my_list[0]
    max_eded = my_list[0]
    
    for i in my_list:
        if max_eded < i:
            max_eded = i
        if min_eded > i :
            min_eded = i
    return f"Maxsimum eded : {max_eded}  Minum eded: {min_eded}"      
print(min_max([3, -1, 5, -7, 0, 2]))
```

```python
def furthestDistanceFromOrigin(moves):
    left = moves.count("L")
    right = moves.count("R")
    blank = moves.count("_")
    
    return abs(right - left) + blank

print(furthestDistanceFromOrigin("_R__LL_"))
```

```python
def mergeKLists(lists):
    my_list = []
    
    for i in lists:
        my_list +=i
    my_list.sort()
    return my_list
print(mergeKLists(lists = [[1,4,5],[1,3,4],[2,6]]))
```

```python
def recoverOrder (order , friends):
    my_list=[]
    
    for item in order:
        if item in friends:
            my_list.append(item)
    return my_list
print(recoverOrder( order = [3,1,2,5,4], friends = [1,3,4]))
```

```python
def earliestTime(tasks):
    min_time = sum(tasks[0])
    for item in tasks:
        if min_time > sum(item):
            min_time = sum(item)
    return min_time
print(earliestTime([[1,6],[2,3]]))   
```

```python
def firstMatchingIndex(s):
    n = len(s)
    for i in range(n):
        if s[i] == s[n - i - 1]:
            return i
    return -1
    
print(firstMatchingIndex("abcacbd"))
```

```python
def reverseByType(s):
    letters = []
    characters = []
    new_word = []
    [letters.append(i) if i.isalpha() else characters.append(i) for i in s]
    
    letters.reverse()
    characters.reverse()
    l=0
    c=0
    
    for i in s:
        if i.isalpha():
            new_word.append(letters[l])
            l +=1
        else:
            new_word.append(characters[c])
            c +=1
    return "".join(new_word)
               
    
print(reverseByType(")ebc#da@f("))
```

```python
def reversePrefix(s, k):
    return s[:k][::-1] + s[k:]

print(reversePrefix(s="abcd", k=2))
```

```python
def minMoves(nums):
    max_m = max(nums)
    total = 0
    for item in nums:
        total += max_m - item
    return total
print(minMoves([4,4,5]))
```

```python
def findMissingElements(nums):
    find_item = []
    max_item = max(nums)
    min_item = min(nums)
    ferq = max_item - min_item + 1
    
    for i in range(ferq):
        if max_item != min_item:
            find_item.append(min_item)
            min_item += 1
    find_item.append(max_item)
    
    s =[]
    
    for i in find_item:
        if i  not in  nums:
            s.append(i)
    return s     

print(findMissingElements([1,4,2,5]))
```

```python
def alternatingSum(nums):
    total = 0
    
    for i in range(len(nums)):
        if i % 2 == 0:
            total += nums[i]
        else:
            total -= nums[i]
    return total

print(alternatingSum([1,3,5,7]))
```

```python
def twoSum(nums, target):
    for i, item in enumerate(nums):
        s = target - item
        if s in nums:
            a = i
            b = nums.index(s)
            if a != b:
                return [b, a]     
            
print(twoSum(nums = [3,3], target = 6))
```

```python
def addTwoNumbers (l1 ,l2):
    l1.reverse()
    l2.reverse()
    l1_str = "".join([str(i) for i in l1])
    l2_str = "".join([str(i) for i in l2])
    
    l3_str = str(int(l1_str) + int(l2_str))
    l3 = [int(i) for i in l3_str]
    l3.reverse()
    return  l3
    
print(addTwoNumbers(l1 = [9,9,9,9,9,9,9], l2 = [9,9,9,9]))
```

```python
def lengthOfLastWord(s):
    a = s.split()
    return len(a[-1])
print(lengthOfLastWord("   fly me   to   the moon  "))
```

```python
def plusOne(digits):
    digits_str = "".join([str(i) for i in digits])
    l3_str = str(int(digits_str) + 1)
    l3 = [int(i) for i in l3_str]
    return  l3
print(plusOne([9]))
```

```python
def findMedianSortedArrays(nums1, nums2):
    nums3 = nums1 + nums2
    nums3.sort()
    mdn = 0
    n = len(nums3)
    if n % 2 == 1:
        mdn = nums3[n//2]
    else:
        mdn = (nums3 [n //2 -1] + nums3[n // 2] ) / 2
    return mdn    
    
print(findMedianSortedArrays(nums1 = [1,3], nums2 = [2]))
```
