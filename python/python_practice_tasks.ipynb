#### 2)  İstifadəçidən iki ədəd və əməliyyat işarəsi (+, -, *, /) daxil etməsini istəyin (məsələn, 5+6). Daxil edilən əməliyyata görə hesablama aparan kalkulyator proqramı yazın.

```python
a = int(input("Birinci ədədi daxil edin: "))
c = input("Əməliyyatı daxil edin (+, -, *, /): ")
b = int(input("İkinci ədədi daxil edin: "))

if c == "+":
    print(f'{a} + {b} = {a + b}')
elif c == "-":
    print(f'{a} - {b} = {a - b}')
elif c == "*":
    print(f'{a} * {b} = {a * b}')
elif c == "/":
    if b != 0:
        print(f'{a} / {b} = {a / b}')
    else:
        print("0-a bölmək olmaz.")
else:
    print("Yanlış əməliyyat daxil etdiniz.")
```

#### 8) İçində siyahılar olan bir siyahı verilib (2D siyahı). max() funksiyasından istifadə etmədən ən böyük ədədi tapan proqram yazın.

```python
def maxNum(lst):
    max_num = lst[0][0]

    for i in lst:
        for j in i:
            if j > max_num:
                max_num = j
    return max_num   
 
maxNum([[1,4,2], [5,11,7], [9,0,-1]])  
```

#### 9) x və y koordinatları olan DataPoint adlı sinif yaradın. Bu sinif başqa bir data pointlər ilə arasındakı Evklid məsafəsini hesablaya bilsin.

```python
class DataPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, point):
        return ((self.x - point.x) ** 2 + (self.y - point.y) ** 2) ** 0.5


point1 = DataPoint(2, 3)
point2 = DataPoint(5, 7)

print(point1.distance(point2))
```
